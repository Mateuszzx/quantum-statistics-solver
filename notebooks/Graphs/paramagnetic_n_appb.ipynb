# Magnetic Distribution

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")

from qss.distributions import prepare_2D_lattice_data
import tools.scientific_plot_v2 as sp
## Przeładowanie bilbiotek pod testowanie zmian
#import importlib
#importlib.reload(sp)
plotter = sp.ScientificPlotter()
```

```python
MU_B  = 1 # eV/T

N = 100
dist = ["FD","FLB","NFL", "NFL_REWORKED","NFLB","SSL", "HUBBARD"]
#distro = ["FD","FLB","NFL","NFL_REWORKED", "NFLB", "SSL", "HUBBARD"]
dist = ["FD","NFL", "NFL_REWORKED", "SSL"]
t = -1
#t_prim = 0.25
t_prim = 0.
kBT = [1, 10, 100]
n = [0.8]
H_a = [0]
U = 100.0


kx = np.linspace(-np.pi, np.pi, N, endpoint=False)
ky = np.linspace(-np.pi, np.pi, N, endpoint=False)
e_k, _, _ = prepare_2D_lattice_data(kx, ky, t, t_prim)
```

```python
# 1D Cut along Gamma-X
plotter.plot_energy_distribution_1d(
    e_k=e_k, 
    slice_type="horizontal",
    dim=2, 
    kx=kx, 
    n=np.array(n), 
    kBTs=np.array(kBT), 
    H=H_a[0],
    U = U,
    distributions=dist,
    figsize=(12, 22),
    ylims=(-0.1, 3.1),
    bz_inset_pos=(0.55, 0.68, 0.25, 0.25)
)
label_map = {
    "NFL": "NFL_old",
    "NFL_REWORKED": "NFL",
}
fig = plt.gcf()
for ax in fig.axes:
    handles, labels = ax.get_legend_handles_labels()
    if not handles:
        continue
    new_labels = [label_map.get(lbl, lbl) for lbl in labels]
    ax.legend(handles, new_labels, loc="center left", fontsize=16)

plt.show()
```

```python
# 1D Cut along Diagonal (Gamma-M)
plotter.plot_energy_distribution_1d(
    e_k=e_k, 
    slice_type="diagonal",
    dim=2, 
    kx=kx, 
    n=np.array(n), 
    kBTs=np.array(kBT), 
    H=H_a[0],
    U=U,
    distributions=dist,
    figsize=(12, 22),
    ylims=(-0.1, 3.1),
    bz_inset_pos=(0.55, 0.68, 0.25, 0.25)
)
label_map = {
    "NFL": "NFL_old",
    "NFL_REWORKED": "NFL",
}
fig = plt.gcf()
for ax in fig.axes:
    handles, labels = ax.get_legend_handles_labels()
    if not handles:
        continue
    new_labels = [label_map.get(lbl, lbl) for lbl in labels]
    ax.legend(handles, new_labels, loc="center left", fontsize=16)

plt.show()
```
