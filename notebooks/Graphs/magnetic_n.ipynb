# Magnetic Distribution

```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import numpy as np
import matplotlib.pyplot as plt

from qss.distributions import  prepare_2D_lattice_data

import tools.scientific_plot_v2 as sp
plotter = sp.ScientificPlotter()
```

```python
MU_B  = 1 # eV/T

N = 100
dist = ["FD","FLB","NFL", "NFLB","SSL"]
t = -1
t_prim = 0
beta = [1, 10, 100]
n = [0.8]
H_a = [1e-3]
kx = np.linspace(-np.pi, np.pi, N, endpoint=False)
ky = np.linspace(-np.pi, np.pi, N, endpoint=False)
e_k, _, _ = prepare_2D_lattice_data(kx, ky, t, t_prim)
```

```python
# 1D Cut along Gamma-X
plotter.plot_energy_distribution_1d(
    e_k=e_k, 
    slice_type="horizontal",
    dim=2, 
    kx=kx, 
    n=np.array(n), 
    betas=np.array(beta), 
    H=H_a[0],
    distributions=dist,
    figsize=(20,8),
    ylims=(-0.1, 1.8),
    bz_inset_pos=(0.58, 0.68, 0.25, 0.25)
)
plt.show()
```

```python
# 1D Cut along Diagonal (Gamma-M)
plotter.plot_energy_distribution_1d(
    e_k=e_k, 
    slice_type="diagonal",
    dim=2, 
    kx=kx, 
    n=np.array(n), 
    betas=np.array(beta), 
    H=H_a[0],
    distributions=dist,
    figsize=(20,8),
    ylims=(-0.1, 1.8),
    bz_inset_pos=(0.58, 0.68, 0.25, 0.25)
)
plt.show()
```
