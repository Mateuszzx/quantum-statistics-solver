```python
import sympy as sp
sp.init_printing()
```

```python
# NFLB paramagnetic
nk, x, theta = sp.symbols('nk x theta')
nk_bar, x_bar = sp.symbols('nk_bar x_bar')
expr = sp.Eq(nk**2 * (1-nk)*x , 2*nk * (1-2*nk)**2)
solution = sp.solve(expr, nk)
```
